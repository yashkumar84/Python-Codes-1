```python
#Dataframes

dict = {
    "names" : ["Yash" , "Aryan" , "Aman" , "Arpit"],
    "age" : [20 , 21 , 17 , 23],
    "city" : ["Delhi" , "Ghaziabad" , "Meerut" , "Jaipur"]
}

df = pd.DataFrame(dict)
print(df)
```

```python
print(df[df["city"] == "Delhi"])
```

```python
print(df[df["age"] >= 20])
```

```python
print(df[(df["age"] >= 18) | (df["city"] == "Delhi")])
```

```python
df.loc[df["city"] == "Delhi"]
```

```python
df.iloc[0]
```

```python
print(df[df["city"].isin(["Delhi" , "Jaipur"])])
```

```python
print(df[df["age"].between(18 , 21)])
```
