```python
#Dashboards

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import between

df = pd.read_excel("./stu_data.xlsx")

sns.set_theme(style="whitegrid")

mark_cols = ["Math_Marks" , "Science_Marks" , "English_Marks"]


# Clean The Data

df[mark_cols + ["Attendance_%" , "Study_Hours_Week"]] = df[mark_cols + ["Attendance_%" ,"Study_Hours_Week"]].apply(pd.to_numeric , errors="coerce")

valid = df.dropna(subset=["Math_Marks" ,"Science_Marks" ,"English_Marks" , "Attendance_%" ,"Study_Hours_Week" , "City" , "Gender"  ]).copy()

valid = valid[valid[mark_cols].apply(lambda col:col.between(0 , 100)).all(axis=1) & valid["Attendance_%"].between(0 , 100)  & valid["Study_Hours_Week"].between(0 , 168)].copy()

valid["Average_Marks"] = valid[mark_cols].mean(axis=1 , skipna=False)


valid["Percentage"] = valid["Average_Marks"]

#Calculate KPIS

total_students =len(valid)
avg_marks = valid["Average_Marks"].mean()
average_attendance = valid["Attendance_%"].mean()
```

```python
pivot = pd.pivot_table(valid , index="City" , values="Math_Marks" , aggfunc="mean")
print(pivot)
```

```python
pivot_new = pd.pivot_table(valid[valid["City"].astype(str).str.strip().str.lower() == "delhi"] , index="Name" , values="Math_Marks" , aggfunc="mean")
print(pivot_new)
```

```python
pivot_Gender = pd.pivot_table(valid, index="Gender", values="Math_Marks", aggfunc = "mean")
print(pivot_Gender)
```
