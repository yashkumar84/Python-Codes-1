```python
valid = students.dropna(subset= marks_cols + ["City" , "Attendance_%" , "Study_Hours_Week" , "Extracurricular"]).copy()
```

```python
valid = valid[valid[marks_cols].apply(lambda col : col.between(0 , 100)).all(axis=1) & valid["Attendance_%"].between(0 , 100) & valid["Study_Hours_Week"].between(0 , 100)]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
from cProfile import label


plt.figure(figsize=(10 , 4))

plt.plot(
    valid["Student_ID"].head(20),
    valid["Math_Marks"].head(20),
    marker="o",
    label="Math"
)

plt.title("Math Marks Of First 20 Valid Students")
plt.xlabel("Student_ID")
plt.ylabel("Math_Marks")
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
subject_avg = valid[marks_cols].mean()
```

```python
plt.figure(figsize=(10 , 4))
plt.bar(subject_avg.index , subject_avg.values)

plt.title("Average Marks By Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.ylim(0 , 100)
plt.show()
```

```python
subject_data = valid[marks_cols].melt(var_name="Subject" , value_name="Marks")

sns.barplot(data=subject_data,
x = "Subject",
y = "Marks",
errorbar=None
)

plt.title("Average Marks By Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.ylim(0 , 100)
plt.show()
```
